```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent
DATA_PATH = BASE_DIR / "data" / "raw" / "issues.csv"

issues = pd.read_csv(DATA_PATH)
issues.head()
```

```python
issues.columns
issues.info()
```

```python
issues["issue_priority"].value_counts()
```

```python
issues["issue_priority"].isna().sum()
```

```python
core_cols = [
    "id",
    "issue_priority",
    "issue_type",
    "issue_proj",
    "issue_contr_count",
    "issue_created",
    "issue_resolution_date"
]

issues_core = issues[core_cols]
issues_core.head()
```

```python
issues_core.info()
```

```python
issues_core["issue_priority"] = issues_core["issue_priority"].replace(
    {"unknown": "Medium"}
)
issues_core["issue_priority"].value_counts()
```

```python
urgent_labels = {"High", "Highest", "Blocker"}

issues_core["priority_bin"] = issues_core["issue_priority"].apply(
    lambda x: 1 if x in urgent_labels else 0
)

issues_core["priority_bin"].value_counts()
```

```python
issues_core.head()
```

```python
issues_core["issue_created"] = pd.to_datetime(
    issues_core["issue_created"],
    format="mixed",
    utc=True
)

issues_core["issue_resolution_date"] = pd.to_datetime(
    issues_core["issue_resolution_date"],
    format="mixed",
    utc=True,
    errors="coerce"
)
```

```python
issues_core["resolution_hours"] = (
    issues_core["issue_resolution_date"] - issues_core["issue_created"]
).dt.total_seconds() / 3600

issues_core["created_hour"] = issues_core["issue_created"].dt.hour
issues_core["created_dayofweek"] = issues_core["issue_created"].dt.dayofweek
```

```python
issues_core[["resolution_hours"]].describe()
```

```python
DATA_sample_utterances = BASE_DIR / "data" / "raw" / "sample_utterances.csv"
issues_sample_utterances = pd.read_csv(DATA_sample_utterances)

issues_sample_utterances.head()
```

```python
text_per_issue = (
    issues_sample_utterances
    .fillna({"actionbody": ""})
    .groupby("issueid")["actionbody"]
    .apply(lambda x: " ".join(x.astype(str)))
    .reset_index()
)

text_per_issue.head()
```

```python
issues_full = issues_core.merge(
    text_per_issue,
    left_on="id",
    right_on="issueid",
    how="left"
)

issues_full[["id", "issue_priority", "priority_bin", "actionbody"]].head()
```

```python
utterances = issues_sample_utterances.copy()

utterances = utterances[
    (utterances["author_role"] == "reporter") &
    (utterances["is_private"] == 0)
]

utterances = utterances[["issueid", "actionbody"]]
utterances.head()
```

```python
issue_text = (
    utterances
    .groupby("issueid")["actionbody"]
    .apply(lambda x: " ".join(x.dropna().astype(str)))
    .reset_index()
)

issue_text.head()
```

```python
issues_full = issues_core.merge(
    issue_text,
    left_on="id",
    right_on="issueid",
    how="left"
)

issues_full["actionbody"] = issues_full["actionbody"].fillna("")
issues_full[["actionbody", "priority_bin"]].head()
```

```python
issues_full["actionbody"].str.len().describe()
```

```python
(issues_full["actionbody"].str.len() > 0).mean()
```

```python
issues_full.loc[
    issues_full["actionbody"].str.len() > 0,
    ["actionbody", "priority_bin"]
].head()
```

```python
feature_cols = [
    "issue_contr_count",
    "created_hour",
    "created_dayofweek",
    "resolution_hours"
]

X = issues_full[feature_cols].fillna(0)
y = issues_full["priority_bin"]

X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test)

print(classification_report(y_test, y_pred))
```
